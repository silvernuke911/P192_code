## Phase Shift Profilometry

$$
\phi(x,y) = \tan^{-1}\left(\frac{I_4(x,y)-I_2(x,y)}{I_1(x,y)-I_3(x,y)}\right)
$$

```python
def phase_shift(img1, img2, img3, img4):
    numerator = img4 - img2
    denominator = img1 - img3
    phase = np.arctan2(numerator, denominator)
    return phase
```

```python
import os
import numpy as np
import cv2
from pathlib import Path

def phase_shift(img1, img2, img3, img4):
    """
    Calculate wrapped phase using 4-step phase shift algorithm
    """
    numerator = img4 - img2
    denominator = img1 - img3
    phase = np.arctan2(numerator, denominator)
    return phase

def load_and_process_images(input_dir, output_dir):
    """
    Process all image sets and save phase maps
    """
    input_path = Path(input_dir)
    output_path = Path(output_dir)
    output_path.mkdir(exist_ok=True)
    
    # Group files by type and frequency
    file_groups = {}
    
    # Collect all image files
    for file_path in input_path.glob("*.png"):
        filename = file_path.stem  # Remove extension
        parts = filename.split('p')
        
        if len(parts) == 2:
            base_name = parts[0]  # e.g., 'ballf20', 'cubef30'
            phase_value = parts[1]  # e.g., '0.00', '1.57'
            
            if base_name not in file_groups:
                file_groups[base_name] = {}
            
            file_groups[base_name][phase_value] = str(file_path)
    
    # Process each group
    for base_name, phase_files in file_groups.items():
        # Extract object type and frequency
        if 'f' in base_name:
            # Find the last 'f' to separate type and frequency
            last_f_index = base_name.rfind('f')
            obj_type = base_name[:last_f_index]  # 'ball', 'cube', 'flat'
            frequency = base_name[last_f_index+1:]  # '20', '30', etc.
        else:
            obj_type = base_name
            frequency = "unknown"
        
        print(f"Processing {obj_type} at frequency {frequency}")
        
        # Check if we have all required phase images
        required_phases = ['0.00', '1.57', '3.14', '4.71']
        if all(phase in phase_files for phase in required_phases):
            try:
                # Load images in correct order
                img1 = cv2.imread(phase_files['0.00'], cv2.IMREAD_GRAYSCALE).astype(np.float32)
                img2 = cv2.imread(phase_files['1.57'], cv2.IMREAD_GRAYSCALE).astype(np.float32)
                img3 = cv2.imread(phase_files['3.14'], cv2.IMREAD_GRAYSCALE).astype(np.float32)
                img4 = cv2.imread(phase_files['4.71'], cv2.IMREAD_GRAYSCALE).astype(np.float32)
                
                # Calculate phase map
                phase_map = phase_shift(img1, img2, img3, img4)
                
                # Normalize phase map for visualization (optional)
                phase_vis = cv2.normalize(phase_map, None, 0, 255, cv2.NORM_MINMAX)
                phase_vis = phase_vis.astype(np.uint8)
                
                # Save phase map
                output_filename = f"{obj_type}_f{frequency}_phase.png"
                output_filepath = output_path / output_filename
                
                # Save both raw phase data (as numpy array) and visualization
                np.save(output_path / f"{obj_type}_f{frequency}_phase.npy", phase_map)
                cv2.imwrite(str(output_filepath), phase_vis)
                
                print(f"  Saved: {output_filename}")
                
            except Exception as e:
                print(f"  Error processing {base_name}: {e}")
        else:
            missing = [p for p in required_phases if p not in phase_files]
            print(f"  Missing phases for {base_name}: {missing}")

def main():
    input_directory = "."  # Current directory, change as needed
    output_directory = "IMG_PHASE"
    
    print("Starting phase calculation...")
    load_and_process_images(input_directory, output_directory)
    print("Phase calculation complete!")

if __name__ == "__main__":
    main()
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from skimage.restoration import unwrap_phase

# Directories
phase_dir = r"C:\Users\verci\Documents\code\Physics157\PHYSICS192 CODE\E2A\IMG PHASE"
unwrap_dir = os.path.join(os.path.dirname(phase_dir), "IMG PHASE UNWRAP")

# Create output directory if missing
os.makedirs(unwrap_dir, exist_ok=True)

# List all .npy files
files = [f for f in os.listdir(phase_dir) if f.lower().endswith('.npy')]

print("Unwrapping phase files...\n")

for file in files:
    src_path = os.path.join(phase_dir, file)
    base_name = os.path.splitext(file)[0]
    save_npy = os.path.join(unwrap_dir, f"{base_name}_unwrap.npy")
    save_png = os.path.join(unwrap_dir, f"{base_name}_unwrap.png")

    try:
        # Load wrapped phase
        phase_wrapped = np.load(src_path)

        # Unwrap phase
        phase_unwrapped = unwrap_phase(phase_wrapped)

        # Save unwrapped .npy file
        np.save(save_npy, phase_unwrapped)

        # Save visualization as .png
        plt.figure(figsize=(6, 6))
        plt.imshow(phase_unwrapped, cmap='jet')
        plt.colorbar(label="Unwrapped Phase (radians)")
        plt.title(base_name)
        plt.axis('off')
        plt.tight_layout()
        plt.savefig(save_png, dpi=300, bbox_inches='tight')
        plt.close()

        print(f"✅ Saved: {save_npy} and {save_png}")

    except Exception as e:
        print(f"❌ Error processing {file}: {e}")

print("\nAll done! Unwrapped images saved in:")
print(unwrap_dir)
```

$$
    Z_w(x,y) = \frac{Lp\Delta\phi(x,y)}{2\pi d  + p \Delta \phi(x,y)}
$$

where L is the distance from the camera plane to the reference flat, p is the sinusoid or
(“fringe”) frequency, d is the distance between the projector and the camera, and is
the phase difference between the object phase image and reference flat phase image.

Distance from camera = 23 cm
p is the frequency (reference, f=50 stretches in 20 cm and has 50 peaks.)
\Delta p is taken from ball and cube compared to flat, take one for each frequency

I want to make a height map. take the npy file for each phase, get one z-map for ball and cube for each freq, comparing it to the flat.

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib import cm

science_plot(fontsize=9)

# Directories
unwrap_dir = r"C:\Users\verci\Documents\code\Physics157\PHYSICS192 CODE\E2A\IMG PHASE UNWRAP"
zmap_dir = os.path.join(os.path.dirname(unwrap_dir), "IMG HEIGHT MAP")
os.makedirs(zmap_dir, exist_ok=True)

# Constants (in meters)
L = 0.23       # camera-to-flat distance
d = 0.1        # projector-camera distance
freq_to_p = {20: 100, 30: 150, 40: 200, 50: 250}  # m^-1

# Objects and frequencies
objects = ["ball", "cube"]
freqs = [20, 30]#, 40, 50]

for obj in objects:
    for f in freqs:
        try:
            # Load unwrapped phase maps
            obj_path = os.path.join(unwrap_dir, f"{obj}_f{f}_phase_unwrap.npy")
            flat_path = os.path.join(unwrap_dir, f"flat_f{f}_phase_unwrap.npy")

            phi_obj = np.load(obj_path)
            phi_flat = np.load(flat_path)

            # Phase difference
            dphi = phi_obj - phi_flat

            # Frequency
            p = freq_to_p[f]

            # Compute height map
            Z = (L * p * dphi) / (2 * np.pi * d + p * dphi)

            # Determine color limits from central region (ignore 1/8 edges)
            ny, nx = Z.shape
            y1, y2 = ny // 8, ny - ny // 8
            x1, x2 = nx // 8, nx - nx // 8
            Z_center = Z[y1:y2, x1:x2]
            vmin = np.min(Z_center) * 0.5
            vmax = np.max(Z_center) * 1.5

            # Save .npy
            np.save(os.path.join(zmap_dir, f"{obj}_f{f}_zmap.npy"), Z)

            # --- 2D Height Map ---
            fig, ax = plt.subplots(figsize=(6, 6))
            im = ax.imshow(Z, cmap='inferno')
            ax.set_title(f"{obj} $f$={f}")
            ax.axis('off')

            # Colorbar same height as image
            cbar = fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
            cbar.ax.tick_params(labelsize=8)
            cbar.set_label("Height (m)", rotation=270, labelpad=15)

            plt.tight_layout()
            plt.savefig(os.path.join(zmap_dir, f"{obj}_f{f}_zmap.png"), dpi=300, bbox_inches='tight')
            plt.close(fig)

            # --- 3D Surface Plot ---
            fig = plt.figure(figsize=(6, 6))
            ax = fig.add_subplot(111, projection='3d')

            # Generate coordinate grids
            X, Y = np.meshgrid(np.arange(nx), np.arange(ny))

            surf = ax.plot_surface(
                X, Y, -Z,
                cmap='inferno',
                linewidth=0, antialiased=True
            )
            ax.set_title(f"{obj} $f$={f} (3D Surface)")
            ax.set_xlabel('x')
            ax.set_ylabel('y')
            ax.set_zlabel('Height (m)')
            ax.set_xlim(0,nx)
            ax.set_ylim(0,ny)
            ax.view_init(elev=45, azim=60)

            # Box aspect ratio based on image size and z exaggeration
            ax.set_box_aspect((nx, ny, 1000))

            # Colorbar
            cbar = fig.colorbar(surf, ax=ax, shrink=0.6, aspect=10, pad=0.1)
            cbar.set_label("Height (m)")

            plt.tight_layout()
            plt.savefig(os.path.join(zmap_dir, f"{obj}_f{f}_zmap_3D.png"), dpi=300, bbox_inches='tight')
            plt.close(fig)

            print(f"✅ Saved 2D + 3D height maps for {obj} f={f}")

        except Exception as e:
            print(f"❌ Error processing {obj} f={f}: {e}")

print("\nAll done! Height maps saved in:")
print(zmap_dir)
```
