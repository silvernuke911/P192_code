```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
x = np.linspace(0,1,1000)
y = np.linspace(0,1,1000)
xv , yv = np.meshgrid(x,y,indexing='ij')


frequencies = [20,30, 40, 50]
phases = [n*np.pi/2 for n in range(5)]

for freq in frequencies:
    for phase in phases:
        z = np.cos(2*np.pi*freq*xv + phase)


        plt.figure(figsize=(6,6),dpi=300)
        plt.imshow(z.T, cmap = 'gray')
        plt.axis(False)
        plt.tight_layout()
        plt.savefig(f'f{freq}p{phase:.2f}.png')
        print(f'f{freq}p{phase:.2f}')
        plt.show()

```

```python
import os
import numpy as np

img_path = r'C:\Users\verci\Documents\code\Physics157\PHYSICS192 CODE\E2A\E2A IMG'

# Parameters for naming
types = ['ball','flat','cube']
frequencies = [20, 30, 40, 50]
phases = [n * np.pi / 2 for n in range(5)]

# Collect files (sorted by name)
files = sorted([f for f in os.listdir(img_path) if f.lower().endswith(('.png', '.jpg', '.jpeg'))])

# Make sure we have enough names
total_needed = len(types) * len(frequencies) * len(phases)
if len(files) < total_needed:
    print(f"⚠ Warning: Only {len(files)} files but need {total_needed} names")
    
# Rename loop
i = 0
for typei in types:
    for freq in frequencies:
        for phase in phases:
            if i >= len(files):
                break
            old_name = files[i]
            ext = os.path.splitext(old_name)[1]  # preserve extension (.png, .jpg, etc.)
            new_name = f"{typei}f{freq}p{phase:.2f}{ext}"
            
            old_path = os.path.join(img_path, old_name)
            new_path = os.path.join(img_path, new_name)
            
            os.rename(old_path, new_path)
            print(f"Renamed: {old_name} -> {new_name}")
            i += 1
```
