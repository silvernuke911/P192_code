```python
import numpy as np 
import matplotlib.pyplot as plt 
import pandas as pd 
import math
```

```python
# Data values
nofield = np.array([10.12,10.45,12.30,15.20,11.10, 9.85,14.75,13.60])
withfield = np.array([10.18,10.50,12.42,15.35,11.22, 9.90,14.89,13.71])
# Difference of values, mean, stddev, and t stat
deltaF = withfield - nofield
df = len(deltaF)-1
mu_delta = np.mean(deltaF)
sigma_delta = np.std(deltaF, ddof=1)#np.sqrt(np.sum((deltaF-mu_delta)**2)/(df))
t_stat = (mu_delta - 0) / (sigma_delta / np.sqrt(len(deltaF)))

## T-distribution pdf
def t_distribution_pdf(t, df):
    """t: test statistic, df: degrees of freedom"""
    return (math.gamma((df + 1) / 2) / 
            (np.sqrt(df * np.pi) * math.gamma(df / 2)) * 
            (1 + t**2 / df) ** (-(df + 1) / 2))
## T-distribution cdf
def t_distribution_cdf(t, df, n=1000):
    """n: number of integration points (higher = more accurate)"""
    if t < -100:  # Extreme negative values
        return 0.0
    if t > 100:   # Extreme positive values
        return 1.0
    # Integrate from -infinity to t using Simpson's rule
    # We approximate -infinity with a large negative number
    a = -100       # Lower bound
    b = min(t,100) # Upper bound
    h = (b - a) / n
    x = [a + i * h for i in range(n + 1)]
    y = [t_distribution_pdf(x_i, df) for x_i in x]
    # Simpson's rule
    integral = y[0] + y[-1]
    for i in range(1, n):
        if i % 2 == 0:
            integral += 2 * y[i]
        else:
            integral += 4 * y[i]
    integral *= h / 3
    return integral

def t_test_p_value(t_stat, df, two_tailed=True):
    # Calculate probability up to t_stat
    if t_stat >= 0:
        p_one_tailed = 1 - t_distribution_cdf(t_stat, df)
    else:
        p_one_tailed = t_distribution_cdf(t_stat, df)
    if two_tailed:
        return 2 * p_one_tailed
    else:
        return p_one_tailed

print(f"Delta F     : {deltaF}")
print(f"Mean Delta  : {mu_delta:.3f}")
print(f"Stddev      : {sigma_delta:.3f}")
print(f"t statistic : {t_stat:.3f}")
pval = t_test_p_value(t_stat,df,two_tailed=True)
print(f"p-value     : {pval:.8f}")

def conclusions(pval, alpha=0.05):
    print("=============\n Results \n=============")
    if pval <= alpha:
        print(f"p-value less than the significance threshold a={alpha}. We reject the null")
    else:
        print(f"p-value greater than the significance threshold a={alpha}. We fail to reject the null")
conclusions(pval)
```

```python

## Data points
LaserA = np.array([4.85 , 4.92 , 4.88 , 4.91 , 4.89 , 4.93 ]) 
LaserB = np.array([4.75 , 4.70 , 4.78 , 4.73 , 4.80 , 4.76 ])

## means and stds and stderrs
meanA = np.mean(LaserA)
meanB = np.mean(LaserB)
stdA = np.std(LaserA,ddof=1)
stdB = np.std(LaserB,ddof=1)
nA   = len(LaserA)
nB   = len(LaserB)
stderrA = stdA**2/nA
stderrB = stdB**2/nB

## t-statistic 
t_stat = (meanA - meanB)/np.sqrt(stderrA+stderrB)

## degrees of freedom 
df = (stderrA+stderrB)**2 / ((stderrA)**2/(nA-1)+(stderrB)**2/(nB-1))

## p-value, from functions of Problem 1
pval = t_test_p_value(t_stat,df,two_tailed=True)

# printing stuff
print(f"meanA = {meanA:.3f}, meanB = {meanB:.3f}")
print(f"stdA  = {stdA:.3f}, stdB  = {stdB:.3f}")
print(f"t-statistic = {t_stat:.3f}, df = {df:.3f}")
print(f"p-value     = {pval:.8f}")
conclusions(pval)
```
