```python
import os
import fitz  # PyMuPDF

input_path = r'C:\Users\verci\Documents\code\Physics157\PHYSICS192 CODE\Set B\Sig_adder\input_folder'
output_path = r'C:\Users\verci\Documents\code\Physics157\PHYSICS192 CODE\Set B\Sig_adder\output_folder'
input_file = 'P192.1 G1 TR-E5B.pdf'
signatures_folder = r'C:\Users\verci\Documents\code\Physics157\PHYSICS192 CODE\Set B\Sig_adder\signatures'

# Full paths
input_pdf = os.path.join(input_path, input_file)
output_pdf = os.path.join(output_path, input_file)

# Open the PDF
pdf_document = fitz.open(input_pdf)

# Get all image files from signatures folder
signature_files = []
for file in os.listdir(signatures_folder):
    if file.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.tiff', '.tif')):
        signature_files.append(os.path.join(signatures_folder, file))

# Process each page
for page_num in range(len(pdf_document)):
    page = pdf_document[page_num]
    
    # Get page dimensions (A4 size)
    page_rect = page.rect
    page_width = page_rect.width
    page_height = page_rect.height
    
    # Signature size - 0.75 inch horizontally (54 points = 0.75 inch)
    signature_width = 54  # 0.75 inch in points
    
    # Position at very bottom left
    y_position = 20  # Bottom of the page
    x_position = 20  # Left edge of the page
    
    # Add each signature image
    for signature_file in signature_files:
        try:
            # Get image to maintain aspect ratio
            img = fitz.Pixmap(signature_file)
            img_width = img.width
            img_height = img.height
            aspect_ratio = img_height / img_width
            
            # Calculate height based on 0.75 inch width and aspect ratio
            signature_height = signature_width * aspect_ratio
            
            # Define rectangle for the image at bottom left
            img_rect = fitz.Rect(x_position, 
                               page_height - signature_height - y_position,  # Bottom aligned
                               x_position + signature_width, 
                               page_height - y_position)
            
            # Insert the image
            page.insert_image(img_rect, filename=signature_file)
            
            # Move x position for next signature (no spacing, sequential)
            x_position += signature_width
            
        except Exception as e:
            print(f"Error adding signature {signature_file}: {e}")

# Save the modified PDF
pdf_document.save(output_pdf)
pdf_document.close()

print(f"PDF with signatures saved to: {output_pdf}")
print(f"Added {len(signature_files)} signature images")
```
