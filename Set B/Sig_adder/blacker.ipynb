```python
import cv2
import numpy as np

input_path = r"C:\Users\verci\Documents\code\Physics157\PHYSICS192 CODE\Set B\Sig_adder\signatures\rwen.png"
output_path = r"C:\Users\verci\Documents\code\Physics157\PHYSICS192 CODE\Set B\Sig_adder\signatures\rwen_k.png"

# Load as BGRA to preserve alpha channel
img = cv2.imread(input_path, cv2.IMREAD_UNCHANGED)

# If image has no alpha, add one
if img.shape[2] == 3:
    img = cv2.cvtColor(img, cv2.COLOR_BGR2BGRA)

# Extract alpha channel
alpha = img[:, :, 3]

# Create mask: True where pixel is NOT transparent
non_transparent = alpha > 0

# Set RGB to black where non-transparent
img[non_transparent, 0:3] = [0, 0, 0]

# Save result
cv2.imwrite(output_path, img)

print("Done.")
```
