```python
import matplotlib.pyplot as plt 
import numpy as np 


d = 0.2 
s = 0.5
L = 1.25 
L_w =0.01
Lc = 1.14 
M_s = 0.125
M_c = 0.075

theta1 = 0 
theta2 = np.deg2rad(5)  # 5 degrees rotation for pendulum 2

# Rotation function
def rotate_point(x, y, angle):
    x_rot = x * np.cos(angle) - y * np.sin(angle)
    y_rot = x * np.sin(angle) + y * np.cos(angle)
    return x_rot, y_rot

plt.figure(figsize=(5,5),dpi = 300)
# backstop
Lb, Lw = 0.075, d + 0.2
plt.plot([-Lw/2,Lw/2,Lw/2,-Lw/2,-Lw/2],[0,0,Lb,Lb,0],'k-',lw=1)

# pendulum 1 - draw two bars as instructed
# Top bar: from backstop to top of mass
plt.plot([(-d-L_w)/2, (-d+L_w)/2, (-d+L_w)/2, (-d-L_w)/2, (-d-L_w)/2],[0,0,-Lc + M_s/2, -Lc + M_s/2, 0],'k-',lw=1)
# Bottom bar: from bottom of mass to bar end
plt.plot([(-d-L_w)/2, (-d+L_w)/2, (-d+L_w)/2, (-d-L_w)/2, (-d-L_w)/2],[-Lc - M_s/2, -Lc - M_s/2, -L, -L, -Lc - M_s/2],'k-',lw=1)

# pendulum mass
plt.plot([(-d-M_c)/2, (-d+M_c)/2, (-d+M_c)/2, (-d-M_c)/2, (-d-M_c)/2],[-Lc + M_s/2, - Lc + M_s/2, -Lc - M_s/2, -Lc-M_s/2, -Lc+M_s/2],'k-',lw=1, alpha=1)  

# pendulum 2 bar - ROTATED
# Define pendulum 2 points relative to its pivot at (d/2, 0)
pend2_top_points = np.array([
    [(-L_w)/2, 0],                    # top left
    [(L_w)/2, 0],                     # top right  
    [(L_w)/2, -Lc + M_s/2],           # bottom right (top of mass)
    [(-L_w)/2, -Lc + M_s/2],          # bottom left (top of mass)
    [(-L_w)/2, 0]                     # back to top left
])

pend2_bottom_points = np.array([
    [(-L_w)/2, -Lc - M_s/2],         # top left (bottom of mass)
    [(L_w)/2, -Lc - M_s/2],          # top right (bottom of mass)
    [(L_w)/2, -L],                   # bottom right
    [(-L_w)/2, -L],                  # bottom left
    [(-L_w)/2, -Lc - M_s/2]          # back to top
])

mass_points = np.array([
    [(-M_c)/2, -Lc + M_s/2],
    [(M_c)/2, -Lc + M_s/2],
    [(M_c)/2, -Lc - M_s/2],
    [(-M_c)/2, -Lc - M_s/2],
    [(-M_c)/2, -Lc + M_s/2]
])

# Rotate all points of pendulum 2
pend2_top_rotated = np.array([rotate_point(x, y, theta2) for x, y in pend2_top_points])
pend2_bottom_rotated = np.array([rotate_point(x, y, theta2) for x, y in pend2_bottom_points])
mass_rotated = np.array([rotate_point(x, y, theta2) for x, y in mass_points])

# Translate to actual position (pivot at d/2, 0)
pend2_top_final = pend2_top_rotated + np.array([d/2, 0])
pend2_bottom_final = pend2_bottom_rotated + np.array([d/2, 0])
mass_final = mass_rotated + np.array([d/2, 0])

# Draw pendulum 2 bars
plt.plot(pend2_top_final[:, 0], pend2_top_final[:, 1], 'k-', lw=1)
plt.plot(pend2_bottom_final[:, 0], pend2_bottom_final[:, 1], 'k-', lw=1)
plt.plot(mass_final[:, 0], mass_final[:, 1], 'k-', lw=1, alpha=1)

# Spring - adjusted to connect to rotated pendulum 2
# Find the spring attachment point on rotated pendulum 2
# Original attachment point was at (d/2, -s) relative to pendulum 2's pivot
spring_attach_pend2_x, spring_attach_pend2_y = rotate_point(0, -s, theta2)
spring_attach_pend2_x += d/2  # translate to global coordinates
spring_attach_pend2_y += 0

# Spring connects from (-d/2, -s) to the rotated attachment point
spring_width = 0.02 
start, end = -d/2, spring_attach_pend2_x
spring_length = abs(end - start)
spring = []
winds = 20
for i in range(winds+1):
    if i%2==0:
        sign=1
    else:
        sign=-1
    # Spring follows the line between the two attachment points
    x_pos = -d/2 + i*spring_length/winds
    # Linear interpolation for y position
    y_pos = -s + (spring_attach_pend2_y + s) * (i/winds) + sign*spring_width
    spring.append([x_pos, y_pos])
spring=np.array(spring)
plt.plot(spring[:,0],spring[:,1],'k-',lw=1.25)

# Sensors
sensor_width, sensor_height = 0.075, 0.2
sensor = np.array([[-sensor_width/2,sensor_width/2,sensor_width/2, -sensor_width/2, -sensor_width/2],
                   [-sensor_height/2,-sensor_height/2,sensor_height/2,sensor_height/2,-sensor_height/2]])
sensor_screen = np.array([[-0.02/2,0.02/2,0.02/2, -0.02/2, -0.02/2],
                   [-0.05/2,-0.05/2,0.05/2,0.05/2,-0.05/2]])
sensor_d =0.35
plt.plot(sensor_screen[0]-(sensor_d-0.025), sensor_screen[1]-Lc+0.02,'k-',lw=1)
plt.plot(sensor_screen[0]+(sensor_d-0.025), sensor_screen[1]-Lc+0.02,'k-',lw=1)
plt.plot(sensor[0]-sensor_d,sensor[1] - Lc,'k-',lw=1)
plt.plot(sensor[0]+sensor_d,sensor[1] - Lc,'k-',lw=1)

## guide lines
plt.plot([d/2,d/2],[0,-1.3],'k--',lw=0.75) #pendulum 2 vertical

# Fixed sensor 1 line - proper x measurement line
sensor1_x = -(sensor_d-0.025) - 0.01  # x-position of sensor 1 screen
mass1_left_x = (-d-M_c)/2  # left edge of mass 1
plt.plot([sensor1_x, mass1_left_x], [-Lc+0.02, -Lc+0.02], 'k--', lw=0.75, alpha=0.7)  # sensor 1 x measurement

# Sensor 2 line that disappears upon contact with mass 2
# Calculate the rightmost point of rotated mass 2
mass2_points_x = mass_final[:, 0]
mass2_right_x = np.max(mass2_points_x)
sensor2_x = (sensor_d-0.025) + 0.01  # x-position of sensor 2 screen

# Only draw sensor 2 line if it doesn't intersect with mass 2
if sensor2_x < mass2_right_x:
    # Line would intersect mass 2, so make it transparent
    plt.plot([sensor2_x, d/2+0.25], [-Lc+0.02, -Lc+0.02], 'k--', lw=0.75, alpha=0)  # transparent
else:
    # Line doesn't intersect mass 2, draw it normally
    plt.plot([sensor2_x, d/2+0.13], [-Lc+0.02, -Lc+0.02], 'k--', lw=0.75, alpha=0.7)  # sensor 2 x measurement

# Plot a circle of radius 0.75, centered at -d/2,0
circle_center_x, circle_center_y = d/2, 0
circle_radius = 0.75
theta = np.linspace(0, 2*np.pi, 360)
circle_x = circle_center_x + circle_radius * np.cos(theta)
circle_y = circle_center_y + circle_radius * np.sin(theta)

# Highlight only the arc from pendulum 2 vertical to 5 degrees (where it touches pendulum 2 bar)
# Calculate the angles for the arc
start_angle = -np.pi/2  # vertical downward (270 degrees)
end_angle = -np.pi/2 + theta2  # 5 degrees rotated from vertical

# Create the arc points
arc_theta = np.linspace(start_angle, end_angle, 50)
arc_x = circle_center_x + circle_radius * np.cos(arc_theta)
arc_y = circle_center_y + circle_radius * np.sin(arc_theta)

# Plot the arc
plt.plot(arc_x, arc_y, 'k:', lw=1)

plt.grid(True,lw=0.5)
plt.gca().set_aspect('equal')


# annotations
endcap_width= 0.01
h_label_d = 0.1

# d annotations
plt.plot([-d/2,d/2],[h_label_d,h_label_d],'k-',lw=0.75) # d line
plt.plot([-d/2,-d/2],[h_label_d+endcap_width,h_label_d-endcap_width],'k-',lw=0.75) # end caps
plt.plot([+d/2,+d/2],[h_label_d+endcap_width,h_label_d-endcap_width],'k-',lw=0.75) # end caps
plt.text(0, 0.12, r'$d$', ha='center', va='center') # label

# s annotations - with endcaps
s_label_d = 0.05
plt.plot([-d/2-s_label_d,-d/2-s_label_d],[0,-s],'k-',lw=0.75)       # s line
plt.plot([-d/2-s_label_d-endcap_width,-d/2-s_label_d+endcap_width],[0,0],'k-',lw=0.75) # top endcap
plt.plot([-d/2-s_label_d-endcap_width,-d/2-s_label_d+endcap_width],[-s,-s],'k-',lw=0.75) # bottom endcap
plt.text(-d/2-s_label_d+0.025, -s/2,'$s$',ha='center',va='center') 

# L annotations - with endcaps
L_label_d = 0.075
plt.plot([-d/2-L_label_d,-d/2-L_label_d],[0,-Lc],'k-',lw=0.75)      # L line
plt.plot([-d/2-L_label_d-endcap_width,-d/2-L_label_d+endcap_width],[0,0],'k-',lw=0.75) # top endcap
plt.plot([-d/2-L_label_d-endcap_width,-d/2-L_label_d+endcap_width],[-Lc,-Lc],'k-',lw=0.75) # bottom endcap
plt.text(-d/2-L_label_d-0.025, -Lc/2,'$L$',ha='center',va='center')

# length less annotations
plt.text(-0.1,-1.14,'$M_1$',ha='center',va='center',fontsize=7)

# Fix M_2 - position it at center of mass 2 and rotate 5 degrees
# Calculate center of mass 2
mass2_center_x = np.mean(mass_final[:, 0])
mass2_center_y = np.mean(mass_final[:, 1])
# Rotate the text by 5 degrees to align with the mass
plt.text(mass2_center_x, mass2_center_y, '$M_2$', ha='center', va='center', 
         rotation=np.rad2deg(theta2), rotation_mode='anchor',fontsize=7)
plt.text(0,-0.55,'spring',ha='center',va='center',fontsize=7)
plt.text(-0.35,-1,'sensor',ha='center',va='center',fontsize=7)
plt.text( 0.35,-1,'sensor',ha='center',va='center',fontsize=7)
plt.text(-.25,-1.2,'$x_1$',ha='center',va='center')
plt.text(.25,-1.25,'$x_2$',ha='center',va='center')
plt.text(.135,-0.85,r'$\theta_2$',ha='center',va='center')
plt.grid(False)
plt.axis(False)
plt.show()
```

```python
import matplotlib.pyplot as plt 
import numpy as np 

def science_plot(fontsize=9, scistyle=True, show_latex=True):
    # Default settings (applied to both 2D and 3D)
    if scistyle:
        import scienceplots
        plt.style.use(['science','grid','notebook'])
    if show_latex:
        plt.rcParams.update({
            # Latex Use
            'text.usetex'     : True,        # Use LaTeX for text rendering
            'font.family'     : 'serif',     # Set font family to serif
        })

    plt.rcParams.update({
        # Fontsizes
        'font.size'       : fontsize,    # General font size
        'axes.titlesize'  : fontsize,    # Font size of the axes title
        'axes.labelsize'  : fontsize,    # Font size of the axes labels
        'xtick.labelsize' : fontsize,    # Font size of the x-axis tick labels
        'ytick.labelsize' : fontsize,    # Font size of the y-axis tick labels
        'legend.fontsize' : fontsize,    # Font size of the legend
        'figure.titlesize': fontsize,    # Font size of the figure title

        # Legend
        'legend.fancybox' : False,       # Disable the fancy box for legend
        'legend.edgecolor': 'k',         # Set legend border color to black
    })
science_plot(fontsize=9)

def create_glass():
    """Create glass coordinates"""
    glass_thickness = 0.0075
    glass_width = 0.5
    glass_height = 0.35
    
    glass = np.array([
        [0, glass_width/2, glass_width/2, glass_width/2+glass_thickness, glass_width/2+glass_thickness,
         -(glass_width/2+glass_thickness), -(glass_width/2+glass_thickness), -glass_width/2, -glass_width/2, 0],
        [glass_thickness, glass_thickness, glass_height, glass_height, 0, 0, 
         glass_height, glass_height, glass_thickness, glass_thickness]
    ])
    return glass, glass_width, glass_height, glass_thickness

def create_water_surface(glass_width):
    """Create water surface coordinates"""
    water_height = 0.3
    x = np.linspace(-glass_width/2, glass_width/2, 100)
    water_surface = 0.001 * np.sin(2*np.pi*10*x) + water_height
    return x, water_surface

def create_clamp_stand():
    """Create clamp stand coordinates"""
    clamp_base_width = 0.03
    clamp_base_height = 0.05
    clamp_rod_width = 0.01
    clamp_rod_height = 0.45

    clamp_base = np.array([
        [0, clamp_base_width/2, clamp_base_width/2, -clamp_base_width/2, -clamp_base_width/2, 0],
        [0, 0, clamp_base_height, clamp_base_height, 0, 0]
    ])
    
    clamp_rod = np.array([
        [0, clamp_rod_width/2, clamp_rod_width/2, -clamp_rod_width/2, -clamp_rod_width/2, 0],
        [clamp_base_height, clamp_base_height, clamp_base_height+clamp_rod_height, 
         clamp_base_height+clamp_rod_height, clamp_base_height, clamp_base_height]
    ])
    
    return clamp_base, clamp_rod, clamp_base_width

def create_hanging_rod(glass_height):
    """Create hanging rod coordinates"""
    rod_length = 0.8
    rod_thickness = 0.015
    r_hole = 0.004
    
    theta = np.linspace(0, 2*np.pi, 360)
    hole1 = np.array([r_hole * np.cos(theta), r_hole * np.sin(theta) + glass_height + rod_thickness/2]) 
    hole2 = np.array([r_hole * np.cos(theta) - rod_length/4, r_hole * np.sin(theta) + glass_height + rod_thickness/2]) 
    hole3 = np.array([r_hole * np.cos(theta) + rod_length/4, r_hole * np.sin(theta) + glass_height + rod_thickness/2]) 
    
    hanging_rod = np.array([
        [0, rod_length/2, rod_length/2, -rod_length/2, -rod_length/2, 0],
        [glass_height, glass_height, glass_height+rod_thickness, 
         glass_height+rod_thickness, glass_height, glass_height]
    ])
    
    return hanging_rod, hole1, hole2, hole3, rod_thickness

def create_clamp():
    """Create clamp coordinates"""
    clamp_size = [0.05, 0.04]
    clamp = np.array([
        [0, 0, clamp_size[1], clamp_size[1], 0],
        [0, clamp_size[0], clamp_size[0], 0, 0]
    ])
    return clamp, clamp_size

def create_pendulum(rotation_angle=30):
    """Create pendulum with rotation"""
    pivot = np.array([0, 0.35])
    L = 0.21
    ww, wh = 0.02, 0.05
    
    # Original weight and string
    weight = np.array([
        [0, ww/2, ww/2, -ww/2, -ww/2, 0],
        [pivot[1]-L, pivot[1]-L, pivot[1]-L-wh, pivot[1]-L-wh, pivot[1]-L, pivot[1]-L]
    ])
    string = np.array([[pivot[0], 0], [pivot[1], pivot[1]-L]])

    # Rotation transformation
    rotation_angle_rad = np.radians(rotation_angle)
    rotation_matrix = np.array([
        [np.cos(rotation_angle_rad), -np.sin(rotation_angle_rad)],
        [np.sin(rotation_angle_rad), np.cos(rotation_angle_rad)]
    ])

    # Rotate string
    rotated_string = np.zeros_like(string)
    for i in range(string.shape[1]):
        point = string[:, i] - pivot
        rotated_point = rotation_matrix @ point
        rotated_string[:, i] = rotated_point + pivot

    # Rotate weight
    rotated_weight = np.zeros_like(weight)
    for i in range(weight.shape[1]):
        point = weight[:, i] - pivot
        rotated_point = rotation_matrix @ point
        rotated_weight[:, i] = rotated_point + pivot

    return rotated_string, rotated_weight, pivot

def add_clamp_mask(ax, glass_width, glass_thickness, glass_height, rod_thickness, clamp_size):
    """Add white rectangle to mask objects behind clamp"""
    clamp_x = glass_width/2 + glass_thickness
    clamp_y = glass_height + rod_thickness/2 - clamp_size[1]/2
    
    # Add white rectangle slightly smaller than clamp to hide objects behind it
    mask_padding = 0.0
    ax.fill_between(
        [clamp_x - mask_padding, clamp_x + clamp_size[1] + mask_padding],
        clamp_y - mask_padding,
        clamp_y + clamp_size[0] + mask_padding,
        color='white', zorder=10
    )
    
    # Add white rectangle for left clamp
    ax.fill_between(
        [-clamp_x - clamp_size[1] - mask_padding, -clamp_x + mask_padding],
        clamp_y - mask_padding,
        clamp_y + clamp_size[0] + mask_padding,
        color='white', zorder=10
    )

def add_angle_indicator(ax, pivot, angle_degrees):
    """Add angle indicator lines and arc"""
    # Vertical reference line
    ax.plot([pivot[0], pivot[0]], [pivot[1], pivot[1]-0.3], 'k--', lw=0.75, zorder=1)
    
    # Angle arc
    theta = np.linspace(0, np.radians(angle_degrees), 30)
    ax.plot(0.15 * np.sin(theta), -0.15 * np.cos(theta) + pivot[1], 'k--', lw=0.75, zorder=1)

def add_line_with_endcaps(ax, line, cap_width=0.05, color='r', linewidth=0.75):
    """Add a line with perpendicular endcaps at both ends"""
    # Extract line endpoints
    x1, x2 = line[0]
    y1, y2 = line[1]
    
    # Calculate line direction vector
    dx = x2 - x1
    dy = y2 - y1
    line_length = np.sqrt(dx**2 + dy**2)
    
    # Normalize direction vector
    if line_length > 0:
        dx /= line_length
        dy /= line_length
    else:
        dx, dy = 1, 0
    
    # Calculate perpendicular vector (rotated 90 degrees)
    perp_dx = -dy
    perp_dy = dx
    
    # Calculate cap endpoints for both ends
    cap_half_width = cap_width / 2
    
    # First end cap
    cap1_x1 = x1 - perp_dx * cap_half_width
    cap1_y1 = y1 - perp_dy * cap_half_width
    cap1_x2 = x1 + perp_dx * cap_half_width
    cap1_y2 = y1 + perp_dy * cap_half_width
    
    # Second end cap
    cap2_x1 = x2 - perp_dx * cap_half_width
    cap2_y1 = y2 - perp_dy * cap_half_width
    cap2_x2 = x2 + perp_dx * cap_half_width
    cap2_y2 = y2 + perp_dy * cap_half_width
    
    # Plot the main line
    ax.plot(line[0], line[1], color=color, lw=linewidth)
    
    # Plot the endcaps
    ax.plot([cap1_x1, cap1_x2], [cap1_y1, cap1_y2], color=color, lw=linewidth)
    ax.plot([cap2_x1, cap2_x2], [cap2_y1, cap2_y2], color=color, lw=linewidth)

def main():
    # Create figure
    plt.figure(figsize=(5, 5), dpi=300)
    ax = plt.gca()
    
    # Create components
    glass, glass_width, glass_height, glass_thickness = create_glass()
    x, water_surface = create_water_surface(glass_width)
    clamp_base, clamp_rod, clamp_base_width = create_clamp_stand()
    hanging_rod, hole1, hole2, hole3, rod_thickness = create_hanging_rod(glass_height)
    clamp, clamp_size = create_clamp()
    rotated_string, rotated_weight, pivot = create_pendulum(rotation_angle=30)
    
    # Plot glass and water
    plt.plot(glass[0], glass[1], 'k-', lw=1)
    plt.plot(x, water_surface, 'k-', lw=1)
    
    # Plot clamp stands
    stand_offset = clamp_base_width/2 + glass_width/2 + glass_thickness
    plt.plot(clamp_base[0] + stand_offset, clamp_base[1], 'k-', lw=1)
    plt.plot(clamp_rod[0] + stand_offset, clamp_rod[1], 'k-', lw=1)
    plt.plot(clamp_base[0] - stand_offset, clamp_base[1], 'k-', lw=1)
    plt.plot(clamp_rod[0] - stand_offset, clamp_rod[1], 'k-', lw=1)
    
    # Plot hanging rod and holes
    plt.plot(hanging_rod[0], hanging_rod[1], 'k-', lw=1)
    plt.plot(hole1[0], hole1[1], 'k-', lw=1)
    plt.plot(hole2[0], hole2[1], 'k-', lw=1)
    plt.plot(hole3[0], hole3[1], 'k-', lw=1)
    
    # Add clamp mask before plotting clamps
    add_clamp_mask(ax, glass_width, glass_thickness, glass_height, rod_thickness, clamp_size)
    
    # Plot clamps (on top of mask)
    clamp_x_offset = glass_width/2 + glass_thickness
    clamp_y_offset = glass_height + rod_thickness/2 - clamp_size[1]/2
    plt.plot(clamp[0] + clamp_x_offset, clamp[1] + clamp_y_offset, 'k-', lw=1, zorder=11)
    plt.plot(-clamp[0] - clamp_x_offset, clamp[1] + clamp_y_offset, 'k-', lw=1, zorder=11)
    
    # Plot pendulum
    plt.plot(rotated_string[0], rotated_string[1], 'k-', lw=1)
    plt.plot(rotated_weight[0], rotated_weight[1], 'k-', lw=1)
    
    ## annotations
    l_length = 0.21+0.05/2
    deg30 = np.deg2rad(30)
    l_line = np.array([[0,l_length*np.sin(deg30)],[0.35,0.35-l_length*np.cos(deg30)]])
    l_dist = 0.05/2

    # Use the function to plot line with endcaps
    red_line = np.array([l_line[0]+l_dist*np.cos(deg30), l_line[1]+l_dist*np.sin(deg30)])
    add_line_with_endcaps(ax, red_line, cap_width=0.015, color='k', linewidth=0.75)

    # Add angle indicator
    add_angle_indicator(ax, pivot, 30)
    
    plt.text(0.1,0.275,r'$\ell$',ha='center',va='center',fontsize=12)
    plt.text(0.05,0.175,r'$\theta$',ha='center',va='center',fontsize=12)
    plt.text(0.14,0.1,r'$m$',ha='center',va='center',fontsize=12)
    plt.text(0.06,0.1,r'$\omega$',ha='center',va='center',fontsize=12)
    
    theta = np.linspace(np.deg2rad(5),np.deg2rad(27))
    plt.plot(l_length*np.sin(theta),-l_length*np.cos(theta)+0.35,'k-',lw=0.75)
    xarrow = (l_length*np.sin(theta))[0]
    yarrow = (-l_length*np.cos(theta)+0.35)[0]
    arrow_length = 0.01
    deg5 = np.deg2rad(5)
    arrow_angle = np.deg2rad(20)
    plt.plot([xarrow,xarrow+arrow_length*np.cos(arrow_angle+deg5)],[yarrow, yarrow+arrow_length*np.sin(arrow_angle+deg5)], 'k-',lw=0.75)
    plt.plot([xarrow,xarrow+arrow_length*np.cos(-arrow_angle+deg5)],[yarrow, yarrow+arrow_length*np.sin(-arrow_angle+deg5)],'k-',lw=0.75)
    # Final settings
    ax.set_aspect('equal')
    plt.grid(False)
    plt.axis(False)
    plt.show()

if __name__ == "__main__":
    main()
```
