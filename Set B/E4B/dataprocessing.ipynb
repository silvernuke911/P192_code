```python
## ACCESSING VIDEO INFORMATION
import cv2
import os

folder = r"C:\Users\verci\Documents\code\Physics157\PHYSICS192 CODE\Set B\e4b\vids"

# List all .mov files
for filename in os.listdir(folder):
    if filename.endswith(".mov"):
        path = os.path.join(folder, filename)
        print(f"Opening: {path}")
        cap = cv2.VideoCapture(path)

        # Check if the file opened
        if not cap.isOpened():
            print("Error opening file:", filename)
            continue

        # Get video properties
        fps = cap.get(cv2.CAP_PROP_FPS)
        frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
        duration = frames / fps if fps > 0 else 0

        print(f"  FPS: {fps:.2f}")
        print(f"  Frame count: {frames}")
        print(f"  Duration: {duration:.2f} seconds")

        # Close file
        cap.release()
```

```python
## SEPARATING FRAMES

import cv2
import os

# Paths
base_dir = r"C:\Users\verci\Documents\code\Physics157\PHYSICS192 CODE\Set B\e4b"
vids_dir = os.path.join(base_dir, "vids")
frames_root = os.path.join(base_dir, "vidframes")

# Create main frames directory if it doesn't exist
os.makedirs(frames_root, exist_ok=True)

# Iterate through all .mov files in the vids directory
for file in os.listdir(vids_dir):
    if file.lower().endswith(".mov"):
        video_path = os.path.join(vids_dir, file)
        video_name = os.path.splitext(file)[0]  # e.g. "0", "1000", "Unknown"
        save_dir = os.path.join(frames_root, video_name)
        os.makedirs(save_dir, exist_ok=True)

        cap = cv2.VideoCapture(video_path)
        if not cap.isOpened():
            print(f"⚠️ Could not open {file}")
            continue

        frame_count = 0
        print(f"🎞 Extracting frames from {file}...")

        while True:
            ret, frame = cap.read()
            if not ret:
                break
            frame_count += 1
            frame_filename = f"{frame_count:06d}.png"
            frame_path = os.path.join(save_dir, frame_filename)
            cv2.imwrite(frame_path, frame)

        cap.release()
        print(f"✅ {frame_count} frames saved to {save_dir}")

print("All videos processed!")
```

```python
import cv2
import os

# --- Paths ---
base_dir = r"C:\Users\verci\Documents\code\Physics157\PHYSICS192 CODE\Set B\e4b"
vids_dir = os.path.join(base_dir, "vids")
frames_root = os.path.join(base_dir, "vidframes")

# Create main frames directory if it doesn't exist
os.makedirs(frames_root, exist_ok=True)

# --- OPTIONAL: select only one video ---
single_video = "Unknown.mov"  # uncomment and set to the video filename you want to process

# Iterate through all .mov files in the vids directory
for file in sorted(os.listdir(vids_dir)):
    if not file.lower().endswith(".mov"):
        continue

    # If single_video is defined, skip others
    # try:
    if 'single_video' in locals() and file != single_video:
        continue

    video_path = os.path.join(vids_dir, file)
    video_name = os.path.splitext(file)[0]  # e.g. "0", "1000", "Unknown"
    save_dir = os.path.join(frames_root, video_name)
    os.makedirs(save_dir, exist_ok=True)

    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print(f"⚠️ Could not open {file}")
        continue

    frame_count = 0
    print(f"🎞 Extracting frames from {file}...")

    while frame_count < 900:  # only first 900 frames
        ret, frame = cap.read()
        if not ret:
            break

        frame_count += 1
        frame_filename = f"{frame_count:06d}.png"
        print(f'Currently at {frame_filename}',end='\r')
        frame_path = os.path.join(save_dir, frame_filename)
        cv2.imwrite(frame_path, frame)

    cap.release()
    print(f"\n✅ {frame_count} frames saved to {save_dir}")

print("All videos processed!")
```
